## Min Stack

**Topic:** Stacks · **Difficulty:** Medium

Design a stack that supports push, pop, top, and retrieving the minimum element in constant time.


````python
Design a stack class that supports the push, pop, top, and getMin operations.

* MinStack() initializes the stack object.
* void push(int val) pushes the element val onto the stack.
* void pop() removes the element on the top of the stack.
* int top() gets the top element of the stack.
* int getMin() retrieves the minimum element in the stack.

Each function should run in O(1)O(1) time.
Example 1:

Input: ["MinStack", "push", 1, "push", 2, "push", 0, "getMin", "pop", "top", "getMin"]

Output: [null,null,null,null,0,null,2,1]

Explanation:
MinStack minStack = new MinStack();
minStack.push(1);
minStack.push(2);
minStack.push(0);
minStack.getMin(); // return 0
minStack.pop();
minStack.top();    // return 2
minStack.getMin(); // return 1
Constraints:

* -2^31 <= val <= 2^31 - 1.
* pop, top and getMin will always be called on non-empty stacks.
````

### Solution


````python
class MinStack:
    def __init__(self) -> None:
        """Initialize the main stack and its synchronized minimum stack."""
        self.stack: list[int] = []
        self.min_stack: list[int] = []

    def push(self, val: int) -> None:
        """Push val and record the minimum for the resulting stack state."""
        self.stack.append(val)
        if not self.min_stack:
            # The first value is automatically the current minimum.
            self.min_stack.append(val)
        else:
            # Preserve the smaller of val and the previous minimum.
            current_minimum = self.min_stack[-1]
            self.min_stack.append(min(val, current_minimum))

    def pop(self) -> None:
        """Remove the top value and its corresponding minimum state."""
        self.stack.pop()
        self.min_stack.pop()

    def top(self) -> int:
        """Return the top value without removing it."""
        return self.stack[-1]

    def getMin(self) -> int:
        """Return the minimum value currently in the stack."""
        return self.min_stack[-1]
````

### Explanation


````python
Min Stack
1. Setup

We need a stack supporting four operations:

push(val): add val to the top.
pop(): remove the top value.
top(): return the top value.
getMin(): return the minimum value currently stored.

Every operation must run in:

O(1) time
	​


A normal stack handles push, pop, and top in O(1), but finding the minimum by scanning the whole stack would take:

O(n)

So we need to store additional information.

2. Governing principle

Use two synchronized stacks:

stack: stores every value.
min_stack: stores the minimum value corresponding to each stack state.

When a value is pushed, we also push the new minimum onto min_stack.

Therefore, both stacks always have the same length, and:

min_stack[-1]

is always the minimum of all values currently in stack.

3. Algorithm design
push(val)

Push val onto the main stack.

For the minimum stack:

If this is the first value, then val is automatically the minimum.
Otherwise, compare val with the previous minimum.
new minimum=min(val,current minimum)

Push that result onto min_stack.

pop()

Remove the top element from both stacks because both entries represent the same stack state.

top()

Return the last element of the main stack.

getMin()

Return the last element of the minimum stack.

4. Pseudocode
class MinStack:
    initialize:
        stack = empty list
        min_stack = empty list

    push(val):
        append val to stack

        if min_stack is empty:
            append val to min_stack
        else:
            new_minimum = minimum(val, top of min_stack)
            append new_minimum to min_stack

    pop():
        remove top of stack
        remove top of min_stack

    top():
        return top of stack

    getMin():
        return top of min_stack
5. Python solution
class MinStack:
    def __init__(self) -> None:
        """Initialize the main stack and its synchronized minimum stack."""
        self.stack: list[int] = []
        self.min_stack: list[int] = []

    def push(self, val: int) -> None:
        """Push val and record the minimum for the resulting stack state."""
        self.stack.append(val)

        if not self.min_stack:
            # The first value is automatically the current minimum.
            self.min_stack.append(val)
        else:
            # Preserve the smaller of val and the previous minimum.
            current_minimum = self.min_stack[-1]
            self.min_stack.append(min(val, current_minimum))

    def pop(self) -> None:
        """Remove the top value and its corresponding minimum state."""
        self.stack.pop()
        self.min_stack.pop()

    def top(self) -> int:
        """Return the top value without removing it."""
        return self.stack[-1]

    def getMin(self) -> int:
        """Return the minimum value currently in the stack."""
        return self.min_stack[-1]
6. Trace the example

Start with:

stack     = []
min_stack = []
push(1)

The stack is initially empty, so 1 is the minimum.

stack     = [1]
min_stack = [1]
push(2)

Compare the new value with the previous minimum:

min(2,1)=1
stack     = [1, 2]
min_stack = [1, 1]
push(0)
min(0,1)=0
stack     = [1, 2, 0]
min_stack = [1, 1, 0]
getMin()

Return:

min_stack[-1]

Therefore:

0
	​

pop()

Remove the top from both stacks:

stack     = [1, 2]
min_stack = [1, 1]
top()
stack[-1] == 2

Therefore:

2
	​

getMin()
min_stack[-1] == 1

Therefore:

1
	​

7. Why duplicate minimums are necessary

Suppose the input is:

push(2)
push(1)
push(1)

The stacks become:

stack     = [2, 1, 1]
min_stack = [2, 1, 1]

After one pop():

stack     = [2, 1]
min_stack = [2, 1]

The minimum is still 1.

Storing a minimum for every stack state avoids special handling for duplicate minimum values.

8. Complexity analysis

Each operation performs only a constant number of list operations:

Operation	Time
push	O(1)
pop	O(1)
top	O(1)
getMin	O(1)

For n stored elements, both stacks can contain n values:

Space complexity=O(n)
	​

9. Alternative: one stack of pairs

Each stack element can instead store:

(value, minimum_at_this_point)
class MinStack:
    def __init__(self) -> None:
        self.stack: list[tuple[int, int]] = []

    def push(self, val: int) -> None:
        if not self.stack:
            current_minimum = val
        else:
            previous_minimum = self.stack[-1][1]
            current_minimum = min(val, previous_minimum)

        self.stack.append((val, current_minimum))

    def pop(self) -> None:
        self.stack.pop()

    def top(self) -> int:
        return self.stack[-1][0]

    def getMin(self) -> int:
        return self.stack[-1][1]

Both methods have the same complexity. The two-stack version is usually easier to understand initially, while the pair-stack version keeps all state in one data structure.

Recommended solution: synchronized main stack and minimum stack
````

## Daily Temperatures

**Topic:** Stacks (Monotonic Stack) · **Difficulty:** Medium

For each day, find how many days until a strictly warmer temperature appears.


````python
Daily Temperatures
Medium

You are given an array of integers temperatures where temperatures[i] represents the daily temperatures on the ith day.

Return an array result where result[i] is the number of days after the ith day before a warmer temperature appears on a future day. If there is no day in the future where a warmer temperature will appear for the ith day, set result[i] to 0 instead.

Example 1:

Input: temperatures = [30,38,30,36,35,40,28]

Output: [1,4,1,2,1,0,0]

Example 2:

Input: temperatures = [22,21,20]

Output: [0,0,0]

Constraints:

* 1 <= temperatures.length <= 1000
* 1 <= temperatures[i] <= 100
````

### Solution


````python
from typing import List


class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        """
        Return the number of days until a warmer temperature occurs.

        Time complexity: O(n)
        Space complexity: O(n)
        """
        number_of_days: int = len(temperatures)

        # result[i] remains 0 if no warmer future day exists.
        result: List[int] = [0] * number_of_days

        # The stack stores indices of unresolved days.
        # Their temperatures remain non-increasing from bottom to top.
        stack: List[int] = []

        for current_day, current_temperature in enumerate(temperatures):

            # Resolve every previous day whose temperature is lower
            # than the current temperature.
            while (
                stack
                and current_temperature > temperatures[stack[-1]]
            ):
                previous_day: int = stack.pop()

                # The difference in indices is the number of days waited.
                result[previous_day] = current_day - previous_day

            # This day now waits for its own future warmer day.
            stack.append(current_day)

        return result


# Local sanity checks
assert Solution().dailyTemperatures([30, 38, 30, 36, 35, 40, 28]) == [1, 4, 1, 2, 1, 0, 0]
assert Solution().dailyTemperatures([22, 21, 20]) == [0, 0, 0]
assert Solution().dailyTemperatures([30]) == [0]
assert Solution().dailyTemperatures([30, 30, 30]) == [0, 0, 0]
assert Solution().dailyTemperatures([30, 40, 50]) == [1, 1, 0]
````

### Explanation


````python
Daily Temperatures — Monotonic Stack
1. Setup

Let:

temperatures[i] be the temperature on day i.
result[i] be the number of days until a strictly warmer temperature occurs.
stack store indices of days whose next warmer temperature has not yet been found.

We need to compute:

result[i] = j - i

where j > i is the first index satisfying:

temperatures[j] > temperatures[i]

If no such j exists:

result[i] = 0

2. Governing principle: monotonic decreasing stack

We maintain a stack of indices whose temperatures are in non-increasing order from bottom to top.

For example:

stack indices:      [1, 3, 4]
stack temperatures: [38, 36, 35]

When we encounter a temperature warmer than the temperature at the top of the stack, we have found the next warmer day for that previous index.

Why store indices instead of temperatures?

We need both:

The previous temperature for comparison.
The distance between the two days.

If the current day is i and the previous day is previous_day, then:

result[previous_day] = i - previous_day

3. Algorithm
Pseudocode
Create result filled with zeros
Create an empty stack

For each index i and temperature current_temperature:

    While the stack is not empty
    and current_temperature is warmer than the temperature
    at the index on top of the stack:

        Remove previous_day from the stack
        Set result[previous_day] = i - previous_day

    Add i to the stack

Return result

The comparison must use > rather than >= because the problem asks for a warmer temperature, not an equal temperature.

4. Python solution
from typing import List


class Solution:
    def dailyTemperatures(self, temperatures: List[int]) -> List[int]:
        """
        Return the number of days until a warmer temperature occurs.

        Time complexity: O(n)
        Space complexity: O(n)
        """
        number_of_days: int = len(temperatures)

        # result[i] remains 0 if no warmer future day exists.
        result: List[int] = [0] * number_of_days

        # The stack stores indices of unresolved days.
        # Their temperatures remain non-increasing from bottom to top.
        stack: List[int] = []

        for current_day, current_temperature in enumerate(temperatures):

            # Resolve every previous day whose temperature is lower
            # than the current temperature.
            while (
                stack
                and current_temperature > temperatures[stack[-1]]
            ):
                previous_day: int = stack.pop()

                # The difference in indices is the number of days waited.
                result[previous_day] = current_day - previous_day

            # This day now waits for its own future warmer day.
            stack.append(current_day)

        return result

5. Full trace

Given:

temperatures = [30, 38, 30, 36, 35, 40, 28]

Initial state:

result = [0, 0, 0, 0, 0, 0, 0]
stack  = []

Day 0: temperature 30

The stack is empty, so push index 0.

stack indices      = [0]
stack temperatures = [30]
result             = [0, 0, 0, 0, 0, 0, 0]

Day 1: temperature 38

Compare 38 with the temperature at index 0:

38 > 30

Pop index 0.

The number of days waited is:

1 - 0 = 1

Therefore:

result[0] = 1

Push index 1.

stack indices      = [1]
stack temperatures = [38]
result             = [1, 0, 0, 0, 0, 0, 0]

Day 2: temperature 30

Compare 30 with 38:

30 is not > 38

No previous day is resolved. Push index 2.

stack indices      = [1, 2]
stack temperatures = [38, 30]
result             = [1, 0, 0, 0, 0, 0, 0]

Day 3: temperature 36

Compare 36 with the temperature at index 2:

36 > 30

Pop index 2.

3 - 2 = 1

Therefore:

result[2] = 1

Now compare 36 with the new stack top, index 1:

36 is not > 38

Stop popping and push index 3.

stack indices      = [1, 3]
stack temperatures = [38, 36]
result             = [1, 0, 1, 0, 0, 0, 0]

Day 4: temperature 35

Compare 35 with 36:

35 is not > 36

Push index 4.

stack indices      = [1, 3, 4]
stack temperatures = [38, 36, 35]
result             = [1, 0, 1, 0, 0, 0, 0]

Day 5: temperature 40

Compare 40 with the temperature at index 4:

40 > 35

Pop index 4.

5 - 4 = 1
result[4] = 1

Compare 40 with index 3:

40 > 36

Pop index 3.

5 - 3 = 2
result[3] = 2

Compare 40 with index 1:

40 > 38

Pop index 1.

5 - 1 = 4
result[1] = 4

Push index 5.

stack indices      = [5]
stack temperatures = [40]
result             = [1, 4, 1, 2, 1, 0, 0]

Day 6: temperature 28

Compare 28 with 40:

28 is not > 40

Push index 6.

stack indices      = [5, 6]
stack temperatures = [40, 28]
result             = [1, 4, 1, 2, 1, 0, 0]

Indices 5 and 6 remain in the stack because neither has a warmer future day. Their answer correctly remains 0.

Final answer: [1, 4, 1, 2, 1, 0, 0]

6. Complexity analysis

Let n be the number of temperatures.

Each index is:

pushed onto the stack once;
popped from the stack at most once.

Therefore, the total number of stack operations is proportional to 2n:

T(n) = O(n)

The stack can contain at most n indices:

S(n) = O(n)

Time: O(n)
Space: O(n)

Why this is better than brute force

A brute-force method checks every future day for every current day:

for i in range(len(temperatures)):
    for j in range(i + 1, len(temperatures)):
        if temperatures[j] > temperatures[i]:
            result[i] = j - i
            break

In a decreasing list such as:

[100, 99, 98, 97, 96]

the inner loop repeatedly scans almost the entire remaining array, giving:

O(n^2)

The monotonic stack avoids repeating those comparisons and reduces the solution to:

O(n)

Reference: the stack operations and aggregate linear-time reasoning follow standard stack and amortized-analysis principles described in Introduction to Algorithms by Cormen, Leiserson, Rivest, and Stein.

Recommended solution: monotonic decreasing stack of indices (next-greater-element pattern)
````

````python
# Interactive step-through of the monotonic-stack algorithm (Example 1)

temperatures = [30, 38, 30, 36, 35, 40, 28]
result = [0] * len(temperatures)
stack: list[int] = []

print("temperatures =", temperatures)
print("Initial: result =", result, "| stack =", stack)
print("-" * 60)

for current_day, current_temperature in enumerate(temperatures):
    print(f"\nDay {current_day}: temperature {current_temperature}")

    while stack and current_temperature > temperatures[stack[-1]]:
        previous_day = stack.pop()
        days_waited = current_day - previous_day
        result[previous_day] = days_waited
        print(
            f"  Pop {previous_day} "
            f"(temp {temperatures[previous_day]}): "
            f"result[{previous_day}] = {current_day} - {previous_day} = {days_waited}"
        )

    stack.append(current_day)
    stack_temps = [temperatures[i] for i in stack]
    print(f"  Push {current_day}")
    print(f"  stack indices      = {stack}")
    print(f"  stack temperatures = {stack_temps}")
    print(f"  result             = {result}")

print("-" * 60)
print("Final answer:", result)
assert result == [1, 4, 1, 2, 1, 0, 0]
````

## Generate Parentheses

**Topic:** Backtracking · **Difficulty:** Medium

Generate all well-formed parenthesis strings with exactly n pairs.


````python
Generate Parentheses
Medium

Given n pairs of parentheses, write a function to generate all combinations of well-formed parentheses.

Example 1:

Input: n = 3
Output: ["((()))","(()())","(())()","()(())","()()()"]

Example 2:

Input: n = 1
Output: ["()"]

Constraints:

* 1 <= n <= 8
````

### Solution


````python
from typing import List


class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        """Generate every well-formed string containing n pairs of parentheses."""

        result: List[str] = []
        current: List[str] = []

        def backtrack(open_count: int, close_count: int) -> None:
            """
            Build one valid parenthesis string.

            open_count:
                Number of opening parentheses currently used.

            close_count:
                Number of closing parentheses currently used.
            """

            # A complete answer must contain n opening and n closing
            # parentheses, giving a total length of 2 * n.
            if len(current) == 2 * n:
                result.append("".join(current))
                return

            # We may add an opening parenthesis while fewer than n
            # opening parentheses have been used.
            if open_count < n:
                current.append("(")

                backtrack(
                    open_count=open_count + 1,
                    close_count=close_count,
                )

                # Undo the choice so another branch can be explored.
                current.pop()

            # We may close a pair only when an unmatched opening
            # parenthesis currently exists.
            if close_count < open_count:
                current.append(")")

                backtrack(
                    open_count=open_count,
                    close_count=close_count + 1,
                )

                # Undo the choice so another branch can be explored.
                current.pop()

        backtrack(open_count=0, close_count=0)

        return result


# Local sanity checks
assert Solution().generateParenthesis(1) == ["()"]
assert Solution().generateParenthesis(2) == ["(())", "()()"]
assert Solution().generateParenthesis(3) == [
    "((()))",
    "(()())",
    "(())()",
    "()(())",
    "()()()",
]
````

### Explanation


````python
Generate Parentheses — Backtracking
1. Setup

We are given an integer n, representing the number of matching parenthesis pairs.

For example, when

n = 3,

we must place:

3 opening parentheses: (
3 closing parentheses: )

A valid answer must satisfy two rules:

It contains exactly n opening and n closing parentheses.
At every point from left to right, the number of closing parentheses cannot exceed the number of opening parentheses.

For example:

(()())   valid
())(()   invalid

The second string becomes invalid at the third character because it closes more pairs than it has opened.

2. Governing algorithm: backtracking

We use backtracking because the problem asks us to generate every valid combination.

At each recursive step, we choose between:

adding (, provided fewer than n opening parentheses have been used;
adding ), provided there is an unmatched opening parenthesis available.

Define:

open_count = number of ( characters currently used;
close_count = number of ) characters currently used;
current = the partial parenthesis string.

The two governing conditions are:

Add "(" when open_count < n
Add ")" when close_count < open_count

The second condition prevents invalid strings from ever being generated.

3. Algorithm design
Initialize an empty result list.

Define a backtracking function:

    If the current string contains 2n characters:
        Add it to the result.
        Return.

    If fewer than n opening parentheses have been used:
        Add "(".
        Continue recursively.
        Remove "(" to restore the previous state.

    If fewer closing parentheses than opening parentheses have been used:
        Add ")".
        Continue recursively.
        Remove ")" to restore the previous state.

Return the result list.

The removal step is the actual backtracking operation. It restores the partial solution so that another choice can be explored.

4. Python solution
from typing import List


class Solution:
    def generateParenthesis(self, n: int) -> List[str]:
        """Generate every well-formed string containing n pairs of parentheses."""

        result: List[str] = []
        current: List[str] = []

        def backtrack(open_count: int, close_count: int) -> None:
            """
            Build one valid parenthesis string.

            open_count:
                Number of opening parentheses currently used.

            close_count:
                Number of closing parentheses currently used.
            """

            # A complete answer must contain n opening and n closing
            # parentheses, giving a total length of 2 * n.
            if len(current) == 2 * n:
                result.append("".join(current))
                return

            # We may add an opening parenthesis while fewer than n
            # opening parentheses have been used.
            if open_count < n:
                current.append("(")

                backtrack(
                    open_count=open_count + 1,
                    close_count=close_count,
                )

                # Undo the choice so another branch can be explored.
                current.pop()

            # We may close a pair only when an unmatched opening
            # parenthesis currently exists.
            if close_count < open_count:
                current.append(")")

                backtrack(
                    open_count=open_count,
                    close_count=close_count + 1,
                )

                # Undo the choice so another branch can be explored.
                current.pop()

        backtrack(open_count=0, close_count=0)

        return result

5. Trace for n = 3

Start with:

current = ""
open = 0
close = 0

We cannot begin with ) because:

close_count < open_count

would mean:

0 < 0,

which is false.

Therefore, the first character must be (.

A partial recursion tree is:

""
└── "("
    ├── "(("
    │   ├── "((("
    │   │   └── "((()"
    │   │       └── "((())"
    │   │           └── "((()))"
    │   └── "(()"
    │       ├── "(()("
    │       │   └── "(()()"
    │       │       └── "(()())"
    │       └── "(())"
    │           └── "(())("
    │               └── "(())()"
    └── "()"
        └── "()("
            ├── "()(("
            │   └── "()(()"
            │       └── "()(())"
            └── "()()"
                └── "()()("
                    └── "()()()"

Therefore:

[
    "((()))",
    "(()())",
    "(())()",
    "()(())",
    "()()()"
]

6. Detailed state trace for one result

Consider generating "(()())".

Step    Current string    open_count    close_count    Choice
0       ""                0             0              Add (
1       "("               1             0              Add (
2       "(("              2             0              Add )
3       "(()"             2             1              Add (
4       "(()("            3             1              Add )
5       "(()()"           3             2              Add )
6       "(()())"          3             3              Complete

At every step:

close_count <= open_count,

so the partial string remains valid.

7. Why the conditions work
Condition 1
if open_count < n:

This prevents the algorithm from using more than n opening parentheses.

Condition 2
if close_count < open_count:

This ensures a closing parenthesis is added only when there is an unmatched opening parenthesis.

For example, suppose:

current = "()"
open_count = 1
close_count = 1

We cannot add ) because:

1 < 1

is false.

The algorithm must add ( next, preventing the invalid prefix:

())

8. Verification
Test 1
solution = Solution()
print(solution.generateParenthesis(1))

Output:

['()']

Test 2
print(solution.generateParenthesis(2))

Output:

['(())', '()()']

Test 3
print(solution.generateParenthesis(3))

Output:

['((()))', '(()())', '(())()', '()(())', '()()()']

Every result:

has length 2n;
contains exactly n opening parentheses;
contains exactly n closing parentheses;
never closes a parenthesis before it has been opened.

9. Complexity analysis

The number of valid parenthesis combinations is the n-th Catalan number:

C_n = (1 / (n + 1)) * C(2n, n)

Each valid string has length 2n. Therefore, the time complexity is:

O(n * C_n)

because the algorithm constructs C_n strings, each of length 2n.

The recursion depth and temporary current list both contain at most 2n characters:

O(n)

auxiliary space, excluding the returned output.

Recommended solution: backtracking with open/close counts (never generate invalid prefixes)
````

### Visualization


````python
# Interactive step-through of the backtracking process (n = 3)
# Shows current string, recursive call stack, choice, and completed solutions.

n = 3
result: list[str] = []
current: list[str] = []
call_stack: list[str] = []


def backtrack(open_count: int, close_count: int, depth: int = 0) -> None:
    state = f'current="{"".join(current)}" open={open_count} close={close_count}'
    call_stack.append(state)
    indent = "  " * depth
    print(f"{indent}ENTER  {state}")
    print(f"{indent}STACK  {call_stack}")

    if len(current) == 2 * n:
        completed = "".join(current)
        result.append(completed)
        print(f"{indent}COMPLETE -> {completed}")
        print(f"{indent}SOLUTIONS SO FAR -> {result}")
        call_stack.pop()
        print(f"{indent}EXIT   {state}")
        return

    if open_count < n:
        print(f"{indent}CHOICE add '('  (open_count {open_count} < n {n})")
        current.append("(")
        backtrack(open_count + 1, close_count, depth + 1)
        current.pop()
        print(f"{indent}UNDO   '('  now current="{"".join(current)}"")

    if close_count < open_count:
        print(f"{indent}CHOICE add ')'  (close_count {close_count} < open_count {open_count})")
        current.append(")")
        backtrack(open_count, close_count + 1, depth + 1)
        current.pop()
        print(f"{indent}UNDO   ')'  now current="{"".join(current)}"")

    call_stack.pop()
    print(f"{indent}EXIT   {state}")


print("=" * 60)
print(f"Generate Parentheses visualization for n = {n}")
print("=" * 60)
backtrack(0, 0)
print("=" * 60)
print("Final solutions:", result)
assert result == ["((()))", "(()())", "(())()", "()(())", "()()()"]
````

## Largest Rectangle in Histogram

**Topic:** Stacks (Monotonic Stack) · **Difficulty:** Hard

Find the largest rectangular area that can be formed in a histogram of unit-width bars.


````python
Largest Rectangle In Histogram
Hard

You are given an array of integers heights where heights[i] represents the height of a bar. The width of each bar is 1.

Return the area of the largest rectangle that can be formed among the bars.

Note: This chart is known as a histogram.

Example 1:

Input: heights = [7,1,7,2,2,4]

Output: 8

Example 2:

Input: heights = [1,3,7]

Output: 7

Constraints:

* 1 <= heights.length <= 100,000
* 0 <= heights[i] <= 10,000
````

### Solution


````python
from typing import List


class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        """Return the largest rectangular area in the histogram."""

        # Each stack entry stores:
        # (earliest_index_where_height_can_start, height)
        stack: List[tuple[int, int]] = []

        max_area = 0

        for i, height in enumerate(heights):
            # Unless taller bars are removed, this bar begins at i.
            start = i

            # A shorter incoming bar establishes the right boundary
            # of every taller bar currently on top of the stack.
            while stack and stack[-1][1] > height:
                previous_start, previous_height = stack.pop()

                # The previous bar can span:
                # previous_start, ..., i - 1
                width = i - previous_start
                area = previous_height * width

                max_area = max(max_area, area)

                # Since the taller bar was able to reach previous_start,
                # the shorter current bar can also extend that far left.
                start = previous_start

            stack.append((start, height))

        n = len(heights)

        # Bars still in the stack never encountered a shorter bar,
        # so their rectangles extend through the end of the histogram.
        for start, height in stack:
            width = n - start
            area = height * width

            max_area = max(max_area, area)

        return max_area


# Local sanity checks
assert Solution().largestRectangleArea([7, 1, 7, 2, 2, 4]) == 8
assert Solution().largestRectangleArea([1, 3, 7]) == 7
assert Solution().largestRectangleArea([5]) == 5
assert Solution().largestRectangleArea([0, 0, 0]) == 0
assert Solution().largestRectangleArea([2, 4, 1]) == 4
````

### Explanation


````python
Largest Rectangle in Histogram — Monotonic Stack
1. Setup

Let:

heights[i] = height of bar i
every bar have width 1
h = rectangle height
w = rectangle width

The area of any rectangle is:

A = h * w

The difficulty is determining the maximum width available to each height.

For heights = [7, 1, 7, 2, 2, 4], the answer is 8 because the bars at indices 2 through 5:

[7, 2, 2, 4]

have minimum height 2 and width 4:

A = 2 * 4 = 8

2. Why brute force is too slow

A naive solution examines every possible interval [i, j].

There are about n(n+1)/2 intervals, requiring O(n^2) time.

Since n <= 100000, that could require about 10^10 operations — far too slow.

We want O(n).

3. Governing principle: Monotonic increasing stack

We maintain a stack containing:

(start_index, height)

The heights in the stack remain monotonically increasing.

Example:

[(0, 1), (2, 2), (5, 4)]

The crucial observation:

When we encounter a bar shorter than the bar on top of the stack, the taller bar can no longer extend to the right.

That means its maximum rectangle can now be calculated.

4. Core idea

Suppose we have [2, 4] and then encounter 1.

The bar of height 4 cannot continue through the 1, so its right boundary has been discovered.

If height 4 started at index 1 and we are currently at index 2:

w = 2 - 1 = 1
A = 4 * 1 = 4

Then we pop 4. Height 2 also cannot continue through 1, so we calculate its rectangle too.

5. Why we store start_index

Consider heights = [2, 4, 1]

Initially:
index 0: push (0, 2)
index 1: push (1, 4)

Stack: [(0, 2), (1, 4)]

At index 2, height = 1:
pop (1, 4) -> area = 4 * (2 - 1) = 4
pop (0, 2) -> area = 2 * (2 - 0) = 4

Now the new bar 1 can extend all the way back to index 0.

Therefore we push (0, 1), not (2, 1).

This is the most important subtlety in the algorithm.

6. Algorithm

For every bar at index i with height h:

    Assume the bar starts at i.
    While the stack contains a bar taller than h:
        pop that bar
        calculate its rectangle
        inherit its starting position
    Push (start, h).

After processing all bars, every remaining bar can extend to the end of the histogram.

7. Pseudocode
max_area = 0
stack = []

for each index i and height h:
    start = i

    while stack is not empty and stack top height > h:
        old_start, old_height = pop stack
        width = i - old_start
        area = old_height * width
        update max_area
        start = old_start

    push (start, h)

for each remaining (start, height):
    width = n - start
    area = height * width
    update max_area

return max_area

8. Trace the example

heights = [7, 1, 7, 2, 2, 4]

Index 0, height = 7
Push (0, 7)
Stack: [(0, 7)]

Index 1, height = 1
7 > 1, pop (0, 7)
width = 1 - 0 = 1, area = 7
max_area = 7
1 inherits index 0
Push (0, 1)
Stack: [(0, 1)]

Index 2, height = 7
1 < 7, no popping
Push (2, 7)
Stack: [(0, 1), (2, 7)]

Index 3, height = 2
7 > 2, pop (2, 7)
width = 3 - 2 = 1, area = 7
start = 2
Push (2, 2)
Stack: [(0, 1), (2, 2)]

Index 4, height = 2
Top height is also 2.
Condition is stack[-1][1] > height (strict), so nothing pops.
Push (4, 2)
Stack: [(0, 1), (2, 2), (4, 2)]

Index 5, height = 4
2 < 4, push (5, 4)
Stack: [(0, 1), (2, 2), (4, 2), (5, 4)]

9. Flush the remaining stack

n = 6. Remaining rectangles extend through index 5.

(0, 1): width = 6, area = 6
(2, 2): width = 4, area = 8  -> max_area = 8
(4, 2): width = 2, area = 4
(5, 4): width = 1, area = 4

Answer: 8

10. Visual interpretation

heights = [7, 1, 7, 2, 2, 4]

                 #
  #       #      #
  #       #      #
  #       #      #
  #       #      #
  #       # # #  #
  # #     # # #  #
  -----------------
    0  1  2 3 4  5

Winning rectangle starts at index 2: [7, 2, 2, 4]
minimum height = 2, width = 4, area = 8

11. Why the algorithm is O(n)

Each bar is pushed once and popped at most once.

Total stack operations proportional to 2n:

T(n) = O(n)
S(n) = O(n)

12. Verification

[7, 1, 7, 2, 2, 4] -> 8
[1, 3, 7] -> 7  (flush: 1*3=3, 3*2=6, 7*1=7)
[5] -> 5
[0, 0, 0] -> 0

13. Pattern to remember

Monotonic stack + next smaller element.

When you see:
- nearest smaller / greater value
- how far an element can extend
- rectangle span
- contribution of each element as a minimum/maximum

consider a monotonic stack.

Invariant:
Stack stores heights that have not yet found their right boundary.
A shorter bar closes every taller rectangle above it.

Recommended solution: monotonic increasing stack of (start_index, height)
````

## Merge K Sorted Linked Lists

**Topic:** Heaps / Priority Queues · Linked Lists · **Difficulty:** Hard

Merge k ascending linked lists into one sorted linked list using a min-heap (K-way merge).


````python
Merge K Sorted Linked Lists
Hard

You are given an array of k linked-lists lists, each linked-list is sorted in ascending order.

Merge all the linked-lists into one sorted linked-list and return it.

Example 1:

Input: lists = [[1,4,5],[1,3,4],[2,6]]
Output: [1,1,2,3,4,4,5,6]
Explanation: The linked-lists are:
[
  1->4->5,
  1->3->4,
  2->6
]
merging them into one sorted linked list:
1->1->2->3->4->4->5->6

Example 2:

Input: lists = []
Output: []

Example 3:

Input: lists = [[]]
Output: []

Constraints:

* k == lists.length
* 0 <= k <= 10^4
* 0 <= lists[i].length <= 500
* -10^4 <= lists[i][j] <= 10^4
* lists[i] is sorted in ascending order
* The sum of lists[i].length will not exceed 10^4
````

### Solution


````python
import heapq
from typing import List, Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def __repr__(self) -> str:
        values = []
        node = self
        while node:
            values.append(str(node.val))
            node = node.next
        return "->".join(values)


class Solution:
    def mergeKLists(
        self,
        lists: List[Optional[ListNode]],
    ) -> Optional[ListNode]:
        """Merge k sorted linked lists into one sorted linked list."""

        # Each heap item stores:
        # (node value, list index, actual node)
        # list index is a tie-breaker so Python never compares ListNode objects.
        heap = []

        # Step 1: add the first node from every non-empty list.
        for i, node in enumerate(lists):
            if node is not None:
                heapq.heappush(heap, (node.val, i, node))

        # Dummy node simplifies construction of the answer.
        dummy = ListNode(0)
        tail = dummy

        # Step 2: repeatedly take the smallest available node.
        while heap:
            value, i, node = heapq.heappop(heap)

            # Attach it to the merged linked list.
            tail.next = node
            tail = tail.next

            # Step 3: the next node from this same list becomes a candidate.
            if node.next is not None:
                heapq.heappush(
                    heap,
                    (node.next.val, i, node.next),
                )

        # dummy itself is not part of the answer.
        return dummy.next


def build_list(values):
    dummy = ListNode(0)
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def to_array(node):
    values = []
    while node:
        values.append(node.val)
        node = node.next
    return values


# Local sanity checks
assert to_array(
    Solution().mergeKLists(
        [
            build_list([1, 4, 5]),
            build_list([1, 3, 4]),
            build_list([2, 6]),
        ]
    )
) == [1, 1, 2, 3, 4, 4, 5, 6]
assert to_array(Solution().mergeKLists([])) == []
assert to_array(Solution().mergeKLists([None])) == []
assert to_array(Solution().mergeKLists([build_list([1, 2, 3])])) == [1, 2, 3]
````

### Explanation


````python
Merge K Sorted Linked Lists — Min Heap / Priority Queue
1. Setup

This problem is really asking:

We have k already-sorted linked lists. How can we efficiently keep selecting the smallest available node among them?

The cleanest solution is a min-heap / priority queue.

Define:

k = number of linked lists
N = total number of nodes across all linked lists
lists[i] = head of the ith sorted linked list
heap = structure that always gives us the smallest currently available node
dummy = temporary node placed before the beginning of our answer
tail = last node currently attached to the merged list

For Example 1:

list 0: 1 -> 4 -> 5
list 1: 1 -> 3 -> 4
list 2: 2 -> 6

Goal:

1 -> 1 -> 2 -> 3 -> 4 -> 4 -> 5 -> 6

2. Governing Principle — Min Heap

A min-heap maintains elements so that the smallest element can be removed efficiently.

Python provides this through:

import heapq

Important operations:

heapq.heappush(heap, item)
heapq.heappop(heap)

Both take O(log k) time when there are at most k elements in the heap.

Why only k elements?

We only store the current smallest unprocessed node from each linked list.

For:

1 -> 4 -> 5
1 -> 3 -> 4
2 -> 6

we initially only need:

1, 1, 2

We do not put every node into the heap at once.

3. Algorithm Design

1. Put the first node of every non-empty linked list into a min-heap.
2. Remove the smallest node from the heap.
3. Attach that node to our merged linked list.
4. If that node has a next node, put the next node into the heap.
5. Repeat until the heap is empty.
6. Return the merged list.

4. Important Python Detail

You might think we could store:

(node.val, node)

But if two nodes have the same value:

(1, nodeA)
(1, nodeB)

Python first compares the values (equal), then tries:

nodeA < nodeB

ListNode does not support <, which can cause TypeError.

So instead store:

(node.val, list_index, node)

The list_index gives Python a numeric tie-breaker.

5. Trace Example 1

Input:

L0: 1 -> 4 -> 5
L1: 1 -> 3 -> 4
L2: 2 -> 6

Initial heap (first node of each list):

(1, list 0), (1, list 1), (2, list 2)

Iteration 1: pop 1 from L0
Answer: 1
Push next from L0: 4
heap ~ [1, 2, 4]

Iteration 2: pop 1 from L1
Answer: 1 -> 1
Push 3
heap ~ [2, 3, 4]

Iteration 3: pop 2 from L2
Answer: 1 -> 1 -> 2
Push 6
heap ~ [3, 4, 6]

Iteration 4: pop 3 from L1
Answer: 1 -> 1 -> 2 -> 3
Push next 4
heap ~ [4, 4, 6]

Iteration 5: pop 4
Answer: 1 -> 1 -> 2 -> 3 -> 4
Push 5
heap ~ [4, 5, 6]

Iteration 6: pop 4
Answer: 1 -> 1 -> 2 -> 3 -> 4 -> 4
No next node
heap ~ [5, 6]

Iteration 7: pop 5
Answer: 1 -> 1 -> 2 -> 3 -> 4 -> 4 -> 5

Iteration 8: pop 6
Answer: 1 -> 1 -> 2 -> 3 -> 4 -> 4 -> 5 -> 6

Heap empty. Result: [1,1,2,3,4,4,5,6]

6. Why This Works

At every moment, the heap contains the smallest remaining node from each active linked list.

Because each individual linked list is already sorted, anything after those candidates must be greater than or equal to them.

Therefore the smallest node in the entire remaining collection must be the smallest node currently in the heap.

7. Complexity

N = total number of nodes
k = number of linked lists

Each node enters and leaves the heap once.
At most k nodes are in the heap.
Each heap operation costs O(log k).

Time: O(N log k)
Space: O(k) extra (heap), excluding the output list

8. Edge Cases

lists = []
Nothing pushed. Return None / []

lists = [[]]
Only list is empty (node is None). Heap stays empty. Return []

One list: [[1,2,3]]
Heap simply reconstructs 1 -> 2 -> 3

9. Pattern to Remember

Classic K-way merge.

Whenever you see several already-sorted sequences that must become one sorted sequence, think:

              smallest candidate
                     |
L0 -> 1 -> 4 -> 5 ---|
                     |---> MIN HEAP ---> merged output
L1 -> 1 -> 3 -> 4 ---|
                     |
L2 -> 2 -> 6 --------|

Expose only one candidate per sorted sequence.

A second legitimate solution is divide-and-conquer pairwise merging, also O(N log k), connecting directly to merge sort.

Recommended solution: min-heap of (value, list_index, node)
````

### Visualization


````python
# Interactive step-through of the min-heap K-way merge
# Example: [[1,4,5],[1,3,4],[2,6]]

import heapq


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    dummy = ListNode(0)
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def to_array(node):
    values = []
    while node:
        values.append(node.val)
        node = node.next
    return values


lists = [
    build_list([1, 4, 5]),
    build_list([1, 3, 4]),
    build_list([2, 6]),
]

heap = []
for i, node in enumerate(lists):
    if node is not None:
        heapq.heappush(heap, (node.val, i, node))
        print(f"Push first node from L{i}: value={node.val}")

print("Initial heap values:", sorted(item[0] for item in heap))
print("-" * 60)

dummy = ListNode(0)
tail = dummy
step = 0

while heap:
    step += 1
    value, i, node = heapq.heappop(heap)
    print(f"\nIteration {step}: pop value={value} from L{i}")

    tail.next = node
    tail = tail.next
    print(f"  Merged so far: {to_array(dummy.next)}")

    if node.next is not None:
        heapq.heappush(heap, (node.next.val, i, node.next))
        print(f"  Push next from L{i}: value={node.next.val}")
    else:
        print(f"  L{i} exhausted")

    print(f"  Heap values now: {sorted(item[0] for item in heap)}")

print("-" * 60)
print("Final answer:", to_array(dummy.next))
assert to_array(dummy.next) == [1, 1, 2, 3, 4, 4, 5, 6]
````

## Swap Nodes in Pairs

**Topic:** Linked Lists · **Difficulty:** Medium

Swap every two adjacent nodes by rewiring pointers. Node values must not be changed.


````python
Swap Nodes in Pairs
Medium

Given a linked list, swap every two adjacent nodes and return its head.
Solve it without modifying the values in the list's nodes — only the nodes
themselves may be changed.

Example 1:
Input: head = [1,2,3,4]
Output: [2,1,4,3]

Example 2:
Input: head = []
Output: []

Example 3:
Input: head = [1]
Output: [1]

Example 4:
Input: head = [1,2,3]
Output: [2,1,3]

Constraints:
* The number of nodes in the list is in the range [0, 100].
* 0 <= Node.val <= 100
````

### Solution


````python
from typing import List, Optional


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

    def __repr__(self) -> str:
        values = []
        node = self
        while node:
            values.append(str(node.val))
            node = node.next
        return "->".join(values)


class Solution:
    def swapPairs(self, head: Optional[ListNode]) -> Optional[ListNode]:
        """Swap every two adjacent nodes by rewiring pointers, not values."""
        dummy = ListNode(0, head)
        prev = dummy

        while prev.next and prev.next.next:
            first = prev.next
            second = first.next

            # prev -> first -> second -> next_pair
            # becomes
            # prev -> second -> first -> next_pair
            first.next = second.next
            second.next = first
            prev.next = second

            # first is now the last node of the swapped pair.
            prev = first

        return dummy.next


def build_list(values: List[int]) -> Optional[ListNode]:
    dummy = ListNode(0)
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def to_array(node: Optional[ListNode]) -> List[int]:
    values = []
    while node:
        values.append(node.val)
        node = node.next
    return values


assert to_array(Solution().swapPairs(build_list([1, 2, 3, 4]))) == [2, 1, 4, 3]
assert to_array(Solution().swapPairs(build_list([]))) == []
assert to_array(Solution().swapPairs(build_list([1]))) == [1]
assert to_array(Solution().swapPairs(build_list([1, 2, 3]))) == [2, 1, 3]
assert to_array(Solution().swapPairs(build_list([1, 2]))) == [2, 1]
````

### Explanation


````python
Swap Nodes in Pairs — Python to C++

This problem is pointer manipulation, not values.

Given:

1 -> 2 -> 3 -> 4

we need:

2 -> 1 -> 4 -> 3

We are not allowed to swap values:

head.val, head.next.val = head.next.val, head.val

We must physically rearrange which nodes point to which nodes.

1. Understand the linked-list operation

One pair:

A -> B -> C

should become:

B -> A -> C

Let:

first = A
second = B
next_pair = C

Three pointer connections to establish:

previous -> second
second   -> first
first    -> next_pair

After swapping one pair we still have to connect it to the previous pair.
A dummy node makes every pair identical, including the first one.

2. Governing principle: dummy node + pointer rewiring

Place a temporary node before the real head:

dummy -> 1 -> 2 -> 3 -> 4

The first swap changes the head:

1 -> 2   becomes   2 -> 1

Without a dummy, that first pair is a special case. With a dummy, every
pair is handled the same way. After all swaps:

dummy -> 2 -> 1 -> 4 -> 3

Return dummy.next.

3. Algorithm design

Suppose:

prev -> first -> second -> next_pair

Three operations:

Operation 1 — first skips over second
    first.next = second.next
    1 -------> 3
    2 -------> 3

Operation 2 — second points back to first
    second.next = first
    2 -> 1 -> 3

Operation 3 — previous section connects to second
    prev.next = second
    prev -> 2 -> 1 -> 3

Then move:

    prev = first

Why first? After the swap, first is the last node of the swapped pair:

2 -> 1
     ^
     |
    prev

The next pair begins at prev.next.

4. Pseudocode

create dummy node pointing to head
prev = dummy

while there are at least two nodes after prev:
    first = prev.next
    second = first.next

    first.next = second.next
    second.next = first
    prev.next = second

    prev = first

return dummy.next

Loop condition:

prev.next exists AND prev.next.next exists

That means "at least two nodes remaining."
If only one node remains, leave it alone.

5. Python line-by-line

dummy = ListNode(0, head)
    dummy -> 1 -> 2 -> 3 -> 4
    The dummy value 0 does not matter. Never return dummy itself.

prev = dummy
    prev is the node immediately before the pair we are about to swap.

while prev.next and prev.next.next:
    Need both a first node and a second node.

first = prev.next
second = first.next
    First iteration: first = 1, second = 2

first.next = second.next
    1 -> 2 -> 3  becomes  1 -> 3

second.next = first
    2 -> 1 -> 3

prev.next = second
    dummy -> 2 -> 1 -> 3 -> 4

prev = first
    dummy -> 2 -> 1 -> 3 -> 4
                  ^
                  |
                 prev
    Next pair begins at prev.next, which is node 3.

6. Full trace: [1,2,3,4]

Initial:
dummy -> 1 -> 2 -> 3 -> 4
 ^
 |
prev

Iteration 1
    first = 1, second = 2
    first.next = second.next   ->  1 -> 3
    second.next = first        ->  2 -> 1 -> 3
    prev.next = second         ->  dummy -> 2 -> 1 -> 3 -> 4
    prev = first

Iteration 2
    first = 3, second = 4
    first.next = second.next   ->  3 -> None   (4.next is None)
    second.next = first        ->  4 -> 3 -> None
    prev.next = second         ->  dummy -> 2 -> 1 -> 4 -> 3
    prev = first

Loop ends: not two more nodes.
Return dummy.next  =>  [2,1,4,3]

7. Odd-length example: [1,2,3]

Start: dummy -> 1 -> 2 -> 3
Swap first pair: dummy -> 2 -> 1 -> 3
prev now points to 1.
Only node 3 remains.
prev.next and prev.next.next fails because 3 has no next.
Node 3 stays untouched: [2,1,3]

8. Edge cases

[]
    dummy -> None
    Loop never runs. Return None / []

[1]
    dummy -> 1 -> None
    No second node. Loop never runs. Result: [1]

[1,2]
    One swap. Result: [2,1]

9. Complexity

Let n = number of nodes.

Each node is visited a constant number of times.
T(n) = O(n)

Only dummy, prev, first, second — no structure that grows with n.
S(n) = O(1) extra space.

Optimal for an iterative solution.

10. Translate the same idea to C++

Do not relearn the algorithm. Translate the same pointer relationships.

ListNode* first means first stores the address of a ListNode.
first->next is shorthand for (*first).next.

Dummy in C++ is a stack object, not a pointer:

    ListNode dummy(0, head);
    ListNode* prev = &dummy;

&dummy means "address of dummy."

Loop condition uses short-circuit && so that if prev->next is nullptr,
C++ does not evaluate prev->next->next (which would dereference null).

Same three rewires:

    first->next = second->next;
    second->next = first;
    prev->next = second;

11. The three lines to memorize conceptually

Before:  prev -> A -> B -> C
After:   prev -> B -> A -> C

A.next = C
B.next = A
prev.next = B

In our variables:

first.next = second.next
second.next = first
prev.next = second

That is the whole problem.

The skill being tested is linked-list pointer rewiring.
If you can draw prev -> A -> B -> C and reconstruct the three new arrows,
you have learned the pattern rather than memorized one LeetCode solution.
````

### Visualization


````python
# Interactive step-through of pair swapping
# Example: [1, 2, 3, 4]  ->  [2, 1, 4, 3]


class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def build_list(values):
    dummy = ListNode(0)
    tail = dummy
    for value in values:
        tail.next = ListNode(value)
        tail = tail.next
    return dummy.next


def to_array(node):
    values = []
    while node:
        values.append(node.val)
        node = node.next
    return values


def snapshot(dummy):
    return to_array(dummy.next)


head = build_list([1, 2, 3, 4])
dummy = ListNode(0, head)
prev = dummy

print("Initial:", snapshot(dummy))
print("-" * 60)

iteration = 0
while prev.next and prev.next.next:
    iteration += 1
    first = prev.next
    second = first.next
    print(f"\nIteration {iteration}: first={first.val}, second={second.val}")
    print(f"  Before: {snapshot(dummy)}")

    first.next = second.next
    print(f"  first.next = second.next  ->  {snapshot(dummy)}")
    print("    (second is detached from the dummy chain until prev.next is rewired)")

    second.next = first
    print(f"  second.next = first       ->  {snapshot(dummy)}")

    prev.next = second
    print(f"  prev.next = second        ->  {snapshot(dummy)}")

    prev = first
    print(f"  prev moves to node {prev.val} (end of swapped pair)")

print("-" * 60)
print("Final answer:", snapshot(dummy))
assert snapshot(dummy) == [2, 1, 4, 3]

print("\nOdd-length leftover: [1, 2, 3]")
odd = ListNode(0, build_list([1, 2, 3]))
prev = odd
while prev.next and prev.next.next:
    first = prev.next
    second = first.next
    first.next = second.next
    second.next = first
    prev.next = second
    prev = first
print("Result:", to_array(odd.next))
assert to_array(odd.next) == [2, 1, 3]
````

## Binary Search

**Topic:** Binary Search · **Difficulty:** Easy

Find `target` in a sorted array of distinct integers in O(log n) time. Return its index, or -1 if it is absent.


````python
Binary Search
Easy

You are given an array of distinct integers nums, sorted in ascending order,
and an integer target.

Implement a function to search for target within nums. If it exists, return
its index, otherwise return -1.

The solution must run in O(log n) time.

Example 1:
Input: nums = [-1,0,2,4,6,8], target = 4
Output: 3

Example 2:
Input: nums = [-1,0,2,4,6,8], target = 3
Output: -1

Constraints:
* 1 <= nums.length <= 10000
* -10000 < nums[i], target < 10000
* All the integers in nums are unique.
````

### Solution


````python
from typing import List


class Solution:
    def search(self, nums: List[int], target: int) -> int:
        """Return the index of target in a sorted array, or -1 if absent."""
        left = 0
        right = len(nums) - 1

        while left <= right:
            # Midpoint of the current closed interval [left, right].
            mid = left + (right - left) // 2

            if nums[mid] == target:
                return mid

            if nums[mid] < target:
                # Everything at or left of mid is too small.
                left = mid + 1
            else:
                # Everything at or right of mid is too large.
                right = mid - 1

        return -1


assert Solution().search([-1, 0, 2, 4, 6, 8], 4) == 3
assert Solution().search([-1, 0, 2, 4, 6, 8], 3) == -1
assert Solution().search([-1, 0, 2, 4, 6, 8], -1) == 0
assert Solution().search([-1, 0, 2, 4, 6, 8], 8) == 5
assert Solution().search([5], 5) == 0
assert Solution().search([5], 0) == -1
````

### C++ translation


```cpp
#include <vector>

class Solution {
public:
    int search(vector<int>& nums, int target) {
        int left = 0;
        int right = static_cast<int>(nums.size()) - 1;

        while (left <= right) {
            int mid = left + (right - left) / 2;

            if (nums[mid] == target) {
                return mid;
            }

            if (nums[mid] < target) {
                left = mid + 1;
            } else {
                right = mid - 1;
            }
        }

        return -1;
    }
};
```

Same algorithm as Python. Only the syntax changes:

| Concept | Python | C++ |
|---|---|---|
| Integer | `left = 0` | `int left = 0;` |
| Array | `List[int]` | `vector<int>` |
| Length | `len(nums)` | `nums.size()` |
| Integer division | `// 2` | `/ 2` with integer operands |
| Array access | `nums[mid]` | `nums[mid]` |
| Loop | `while left <= right:` | `while (left <= right)` |

Because both operands of `/` are integers, C++ discards the fractional part, matching Python's `//`.


### Explanation


````python
Binary Search — Python first, then C++

nums is already sorted and all values are distinct. Each comparison
eliminates about half of the remaining array, which is why this runs in
O(log n) rather than O(n).

1. Setup

nums   = sorted array of n distinct integers
target = value we want to find
left   = leftmost index still under consideration
right  = rightmost index still under consideration
mid    = midpoint of the current search interval

Initially:

left = 0
right = n - 1

Example:

nums = [-1, 0, 2, 4, 6, 8]
index    0  1  2  3  4  5

2. Governing principle

Because the array is strictly increasing:

nums[0] < nums[1] < ... < nums[n-1]

comparing target against nums[mid] has exactly three outcomes:

nums[mid] == target
    Found it. Return mid.

nums[mid] < target
    Everything at or left of mid is too small.
    left = mid + 1

nums[mid] > target
    Everything at or right of mid is too large.
    right = mid - 1

That discard step is the whole algorithm.

3. Pseudocode

left = 0
right = length(nums) - 1

while left <= right:
    mid = left + (right - left) // 2

    if nums[mid] == target:
        return mid

    if nums[mid] < target:
        left = mid + 1
    else:
        right = mid - 1

return -1

The loop condition is left <= right, not left < right.
When left == right, one candidate still remains and must be checked.

4. Trace of Example 1

nums = [-1, 0, 2, 4, 6, 8]
target = 4

Initial: left = 0, right = 5

Iteration 1
    mid = 0 + (5 - 0) // 2 = 2
    nums[2] = 2
    2 < 4, so indices 0, 1, 2 cannot contain 4
    left = 3

    [-1, 0, 2, | 4, 6, 8]
                 L     R

Iteration 2
    left = 3, right = 5
    mid = 3 + (5 - 3) // 2 = 4
    nums[4] = 6
    6 > 4, so indices 4 and 5 cannot contain 4
    right = 3

    Only index 3 remains.

Iteration 3
    left = 3, right = 3
    mid = 3
    nums[3] = 4 == target
    return 3

5. Trace of Example 2

nums = [-1, 0, 2, 4, 6, 8]
target = 3

left right mid nums[mid]  decision
  0     5    2      2     go right  (left = 3)
  3     5    4      6     go left   (right = 3)
  3     3    3      4     go left   (right = 2)

After the last comparison:

left = 3
right = 2

left > right, so the interval is empty. Return -1.

6. Why this is O(log n)

Start with n candidates.

After 1 comparison: about n/2 remain
After 2 comparisons: about n/4 remain
After k comparisons: about n / 2^k remain

Stop when about 1 element remains:

n / 2^k = 1
n = 2^k
k = log2(n)

T(n) = O(log n)

Only left, right, and mid are stored, so extra space is O(1).

7. Why not linear search?

for i in range(len(nums)):
    if nums[i] == target:
        return i

Worst case examines every element: O(n).
The problem requires O(log n). Binary search is valid only because the
array is sorted; that ordering tells you which half is safe to discard.

8. Why compute mid this way?

You will also see:

mid = (left + right) // 2

Mathematically equivalent for these bounds.

The robust form:

mid = left + (right - left) // 2

and in C++:

int mid = left + (right - left) / 2;

avoids overflow of left + right on a fixed-width integer when indices
are huge. Here n <= 10000, so overflow cannot happen, but the safer
expression is the habit to keep.

In C++, / on two ints truncates toward zero, matching Python //.
Example: 5 / 2 = 2.

9. The invariant

At the start of every iteration:

If target exists, it lies in [left, right], inclusive.

When nums[mid] < target:

    target > nums[mid]
    every nums[i] for i <= mid is also < target
    so left = mid + 1 is justified

Symmetric argument: right = mid - 1 when nums[mid] > target.

The code is just this invariant implemented.

10. Common mistakes

Mistake 1 — wrong loop condition
    while left < right
    For this closed-interval version use while left <= right.
    left == right still has one candidate.

Mistake 2 — forgetting + 1
    left = mid
    If left = 3, right = 4, mid = 3, then left = mid leaves the
    interval unchanged and can infinite-loop.
    Use left = mid + 1.

Mistake 3 — forgetting - 1
    right = mid
    Same stall. Use right = mid - 1.

Mistake 4 — binary search on an unsorted array
    Comparing against nums[mid] tells you nothing about which half
    can be discarded unless
    nums[0] < nums[1] < ... < nums[n-1]

11. Pattern to keep

while left <= right:
    find mid
    equal        -> return mid
    target > mid -> left  = mid + 1
    target < mid -> right = mid - 1
return -1

This closed-interval pattern reappears in Search in Rotated Sorted Array,
Find Minimum in Rotated Sorted Array, Koko Eating Bananas, Search a 2D
Matrix, and binary-search-on-the-answer problems.
````

### Visualization


````python
# Interactive step-through of binary search
# Example 1: nums = [-1, 0, 2, 4, 6, 8], target = 4  ->  index 3
# Example 2: same array, target = 3                  ->  -1


def binary_search_trace(nums, target):
    left = 0
    right = len(nums) - 1
    step = 0

    print(f"nums = {nums}")
    print(f"target = {target}")
    print("-" * 60)

    while left <= right:
        step += 1
        mid = left + (right - left) // 2
        print(
            f"\nIteration {step}: left={left}, right={right}, "
            f"mid={mid}, nums[mid]={nums[mid]}"
        )

        if nums[mid] == target:
            print(f"  {nums[mid]} == {target}  ->  return {mid}")
            return mid

        if nums[mid] < target:
            print(f"  {nums[mid]} < {target}  ->  left = {mid + 1}")
            left = mid + 1
        else:
            print(f"  {nums[mid]} > {target}  ->  right = {mid - 1}")
            right = mid - 1

        if left <= right:
            print(f"  interval now [{left}, {right}] = {nums[left:right + 1]}")
        else:
            print(f"  interval empty (left={left}, right={right})")

    print("\nNo remaining interval. Return -1")
    return -1


print("Example 1")
result = binary_search_trace([-1, 0, 2, 4, 6, 8], 4)
print("Final answer:", result)
assert result == 3

print("\n" + "=" * 60)
print("Example 2")
result = binary_search_trace([-1, 0, 2, 4, 6, 8], 3)
print("Final answer:", result)
assert result == -1
````
